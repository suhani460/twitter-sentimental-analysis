```python

# ============================================
# TWITTER SENTIMENT ANALYSIS USING BIG DATA
# Step 1: Library Installation and Spark Setup
# ============================================

!pip install -q pyspark pandas matplotlib seaborn

import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

# Create Spark Session
spark = SparkSession.builder \
    .appName("Twitter Sentiment Analysis") \
    .master("local[*]") \
    .config("spark.driver.memory", "4g") \
    .getOrCreate()

# Display project information
print("=" * 50)
print("TWITTER SENTIMENT ANALYSIS - BIG DATA")
print("=" * 50)

print("PySpark Version:", spark.version)
print("Spark Session:", "Successfully Created")
print("Application Name:", spark.sparkContext.appName)

print("\nProject Setup Completed Successfully!")
```

```python

# ============================================
# STEP 2: DOWNLOAD AND LOAD SENTIMENT140 DATASET
# ============================================

import os
import urllib.request
import zipfile

# Dataset download
url = "https://cs.stanford.edu/people/alecmgo/trainingandtestdata.zip"

zip_path = "/content/twitter_dataset.zip"
extract_path = "/content/twitter_data"

print("Downloading Sentiment140 dataset...")

urllib.request.urlretrieve(url, zip_path)

# Extract dataset
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset downloaded and extracted!")

# Find training dataset
dataset_path = os.path.join(
    extract_path,
    "training.1600000.processed.noemoticon.csv"
)

# Define column names
columns = [
    "target",
    "ids",
    "date",
    "flag",
    "user",
    "text"
]

# Load dataset using PySpark
df = spark.read.csv(
    dataset_path,
    header=False,
    inferSchema=False,
    encoding="ISO-8859-1"
)

df = df.toDF(*columns)

# Display dataset information
print("\nTotal Tweets:", df.count())

print("\nDataset Schema:")
df.printSchema()

print("\nFirst 10 Tweets:")
df.select("target", "text").show(10, truncate=70)
```

```python

# ============================================
# STEP 3: DATA CLEANING AND PREPROCESSING
# ============================================

from pyspark.sql.functions import (
    col, lower, regexp_replace, trim, when
)

# Remove duplicate and null records
clean_df = df.dropDuplicates(["ids", "text"])

clean_df = clean_df.dropna(subset=["text", "target"])

# Remove empty tweets
clean_df = clean_df.filter(trim(col("text")) != "")

# Convert text to lowercase
clean_df = clean_df.withColumn(
    "clean_text",
    lower(col("text"))
)

# Remove URLs
clean_df = clean_df.withColumn(
    "clean_text",
    regexp_replace(col("clean_text"), r"http\S+|www\S+", "")
)

# Remove mentions
clean_df = clean_df.withColumn(
    "clean_text",
    regexp_replace(col("clean_text"), r"@\w+", "")
)

# Remove hashtags symbol but preserve the word
clean_df = clean_df.withColumn(
    "clean_text",
    regexp_replace(col("clean_text"), "#", "")
)

# Remove special characters and numbers
clean_df = clean_df.withColumn(
    "clean_text",
    regexp_replace(col("clean_text"), r"[^a-zA-Z\s]", " ")
)

# Remove extra spaces
clean_df = clean_df.withColumn(
    "clean_text",
    trim(regexp_replace(col("clean_text"), r"\s+", " "))
)

# Remove tweets that became empty
clean_df = clean_df.filter(col("clean_text") != "")

# Convert numerical sentiment into labels
clean_df = clean_df.withColumn(
    "sentiment",
    when(col("target") == "4", "Positive")
    .when(col("target") == "0", "Negative")
    .otherwise("Unknown")
)

# Keep only valid sentiment records
clean_df = clean_df.filter(col("sentiment") != "Unknown")

# Display results
print("DATA CLEANING COMPLETED!")
print("Original Records:", df.count())
print("Cleaned Records:", clean_df.count())

print("\nCleaned Tweets:")
clean_df.select(
    "text", "clean_text", "sentiment"
).show(10, truncate=80)

print("\nSentiment Distribution:")
clean_df.groupBy("sentiment").count().show()
```

```python

# ============================================
# STEP 4: COMPLETE SENTIMENT ANALYSIS MODEL
# ============================================

from pyspark.sql.functions import col, when
from pyspark.ml import Pipeline
from pyspark.ml.feature import Tokenizer, StopWordsRemover, HashingTF, IDF
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.evaluation import MulticlassClassificationEvaluator
import builtins

print("Preparing data...")

# Prepare model dataset
model_df = clean_df.select(
    col("clean_text"),
    when(col("sentiment") == "Positive", 1.0)
    .otherwise(0.0).alias("label")
)

# Split dataset
train_data, test_data = model_df.randomSplit(
    [0.8, 0.2], seed=42
)

print("Training records:", train_data.count())
print("Testing records:", test_data.count())

# Create NLP pipeline
tokenizer = Tokenizer(
    inputCol="clean_text",
    outputCol="words"
)

stopwords = StopWordsRemover(
    inputCol="words",
    outputCol="filtered_words"
)

hashingTF = HashingTF(
    inputCol="filtered_words",
    outputCol="raw_features",
    numFeatures=10000
)

idf = IDF(
    inputCol="raw_features",
    outputCol="features"
)

lr = LogisticRegression(
    featuresCol="features",
    labelCol="label",
    maxIter=10,
    regParam=0.01
)

pipeline = Pipeline(stages=[
    tokenizer, stopwords, hashingTF, idf, lr
])

# Train model
print("\nTraining model...")

model = pipeline.fit(train_data)

print("Model trained successfully!")

# Generate predictions
predictions = model.transform(test_data)

print("\nSample predictions:")

predictions.select(
    "clean_text", "label", "prediction"
).show(10, truncate=50)

# Evaluate model
for metric in ["accuracy", "f1",
               "weightedPrecision", "weightedRecall"]:

    evaluator = MulticlassClassificationEvaluator(
        labelCol="label",
        predictionCol="prediction",
        metricName=metric
    )

    score = evaluator.evaluate(predictions)

    print(metric, ":", builtins.round(score, 4))

print("\nSTEP 4 COMPLETED SUCCESSFULLY!")
```
